```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import OneHotEncoder, LabelEncoder
```

```python
train_data = "titanic_train_imputed.csv"
test_data = "titanic_test_imputed.csv"
train_df = pd.read_csv(train_data)
test_df = pd.read_csv(test_data)

target = "survived"
X_train = train_df.drop(columns=target)
y_train = train_df[target]

X_test = test_df.drop(columns = target)
y_test = test_df[target]

print("X Train shape:", X_train.shape)
print("X Test shape:", X_test.shape)
```

```python
X_train.head()
```

```python
# remove redundant columns

X_train = X_train.drop(columns = ["class", "embark_town"])
X_test = X_test.drop(columns = ["class", "embark_town"])
```

```python
X_train.columns.tolist()
```

```python
num_cols = X_train.select_dtypes(np.number).columns.to_list()
num_cols
```

```python
num_cols = ["age", "sibsp", "parch", "fare"]
cat_cols = ["pclass", "sex", "who", "embarked", "adult_male", "deck", "alone"]
```

```python
binary_ordinal_cols = ["sex", "alone", "adult_male"]
multi_class_nominal_cols = ["embarked", "who", "deck"]
```

```python
X_train_enc = X_train.copy()
X_test_enc = X_test.copy()
```

```python
#### Label Encoding
label_encoders = {}

for col in binary_ordinal_cols:
    le = LabelEncoder()
    le.fit(X_train[col])
    label_encoders[col] = le
    X_train_enc[col]= le.transform(X_train[col])
    X_test_enc[col]= le.transform(X_test_enc[col])
```

```python
X_train_enc.head()
```

#### 7. One Hot encoding

```python
ohe = OneHotEncoder(drop='first', handle_unknown="ignore", sparse_output=False)
```

```python
ohe.fit(X_train[multi_class_nominal_cols])
```

```python
train_ohe = ohe.transform(X_train[multi_class_nominal_cols])
test_ohe = ohe.transform(X_train[multi_class_nominal_cols])
```

```python
train_ohe
```

```python
ohe_cols = ohe.get_feature_names_out(multi_class_nominal_cols)
print(ohe_cols)
```

```python
train_ohe = pd.DataFrame(train_ohe, columns = ohe_cols, index = X_train.index)
test_ohe = pd.DataFrame(test_ohe, columns = ohe_cols, index = X_train.index)
```

```python
train_ohe
```

**Remove original multiclass nominal columns and join the necoded**

```python
X_train_enc = X_train_enc.drop(columns=multi_class_nominal_cols)
X_test_enc = X_test_enc.drop(columns = multi_class_nominal_cols)
```

```python
X_train_enc = pd.concat([X_train_enc, train_ohe], axis=1)
X_test_enc = pd.concat([X_test_enc, test_ohe], axis=1)
```

```python
train_enc_df = X_train_enc.copy()
train_enc_df[target] = y_train.values


test_enc_df = X_test_enc.copy()
#test_enc_df[target] = y_test.values


train_enc_df.to_csv("titanic_train_encoded.csv", index = False)
test_enc_df.to_csv("titanic_test_encoded.csv", index = False)
```
